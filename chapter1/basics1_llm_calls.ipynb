```python
# import os 
# from openai import OpenAI
# client = OpenAI(api_key=os.environ.get("OPENAI_API_KEY"))
# chat_completion = client.chat.completions.create(
#     messages = [
#         {"role":"user","content":"tell me a fun fact"}
#     ],
#     model = "gpt-5-mini",
# )
import os
import httpx
from dotenv import load_dotenv
from openai import OpenAI
 
load_dotenv()
 
 
client = OpenAI(
    api_key=os.environ["OPENAI_API_KEY"],
    base_url="https://alertmanager.tecnicasreunidas.es/v1",
    http_client=httpx.Client(verify=False),
)
```

```python
models = client.models.list()
[m.id for m in models]
```

```python
response = client.chat.completions.create(
    messages = [
        {"role":"user","content":"tell me a quick fact"}
    ],
    model = "gpt-5-mini",
)

print(response.choices[0].message.content)
```


api calls using langchain 



```python
from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0,api_key=os.environ["OPENAI_API_KEY"],
    base_url="https://alertmanager.tecnicasreunidas.es/v1",
    http_client=httpx.Client(verify=False),)
llm_openai.invoke("Hey tell me a historical secret")
```

another way to make api calls using langchain initmodels

```python
from langchain.chat_models import init_chat_model
llm_openai = init_chat_model(model="gpt-5-mini", temperature=0,api_key=os.environ["OPENAI_API_KEY"],
    base_url="https://alertmanager.tecnicasreunidas.es/v1",
    http_client=httpx.Client(verify=False),)

llm_openai.invoke("Hi tell me a fact about telangana mahabubnagar").content
```
