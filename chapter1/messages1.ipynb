# *MEssages in Langchain*

```python

from langchain_core.messages import SystemMessage, HumanMessage,AIMessage
from langchain_openai import ChatOpenAI
import os
import httpx
from dotenv import load_dotenv #loads all the variables from the .env gile and make them available in the os.environ  dictionary(env variables)

if os.environ.get("OPENAI_API_KEY"):
    print("api key is exists...")
else:raise ValueError("key is not in the environment")
my_message = [
    SystemMessage(content="You are a gen-z assistant who always answers in a funny gen z way"),
    HumanMessage(content="Bro tell me a wildest fun fact")

]
llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0,api_key=os.environ["OPENAI_API_KEY"],
    base_url="https://alertmanager.tecnicasreunidas.es/v1",
    http_client=httpx.Client(verify=False),)
llm_openai.invoke(my_message).content
```

# **prompts**

prompts are the messages that are sent to the LLM. Prompts are more user-friendly than messages

```python
from langchain_core.prompts import PromptTemplate

user_input =  input("which fun fact you want to know about, enter it!")

dynamic_prompt = PromptTemplate.from_template("Tell me the fun fact about {topic}")

ready_prompt = dynamic_prompt.invoke({"topic": user_input})

llm_openai.invoke(ready_prompt).content
```

Setting the tone of the system...


```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system","You are a polite assistant"),
    ("user","Write a fun fact about {topic}")
])

user_input = input("Enter a topic")
ready_prompt = prompt_template.invoke({"topic":user_input})
llm_openai.invoke(ready_prompt.messages).content

```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ('system',"You are a {tone} assistant"),
    ("user","tell me the fact about {topic}")
])
user_tone =input("Enter the tone you want to choose")
user_topic = input("Enter the topic")
ready_prompt = prompt_template.invoke({'tone':user_tone,'topic':user_topic})
llm_openai.invoke(ready_prompt.messages).content
```

```python
from langchain_openai import ChatOpenAI
from openai.types import model
from langchain_core.prompts import ChatPromptTemplate
import os
from dotenv import load_dotenv
import httpx

llm_connection = ChatOpenAI(model='gpt-5-mini',
                    temperature = 0,
                    api_key=os.environ["OPENAI_API_KEY"],
                    base_url="https://alertmanager.tecnicasreunidas.es/v1",
                    http_client=httpx.Client(verify=False))

prompt_template =  ChatPromptTemplate.from_messages([
    ("system","you are a {tone} assistant"),
    ("user","Get the information for the {topic}")
])

user_tone = input("set the tone in a single word..!\n eg: Professioal or Funny or GenZ or Friendly")
user_input = input("Enter the topic you want the information for.....!")

final_prompt = prompt_template.invoke({"tone":user_tone,"topic":user_input})

llm_answer = llm_connection.invoke(final_prompt.messages).content
llm_answer
```
