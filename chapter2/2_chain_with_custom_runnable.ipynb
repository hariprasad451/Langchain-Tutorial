```python
# Task-4 [Custom Runnable]#we need to make our custom function runnable so we need to use runnable lamda from langchain and create a runnable lamda 

from langchain_core.runnables import RunnableLambda
def dict_maker(text:str)->dict:

    return {"text":text}
#dict1 = dict_maker()
dict_maker_runnable =  RunnableLambda(dict_maker)
dict_maker_runnable
```

```python
#Task -5 [template for post]
from langchain_core.prompts import ChatPromptTemplate
post_prompt = ChatPromptTemplate.from_messages([
    ("system","You are a professional Social media post generator"),
    ("human","create a post for the following user input for linkedIn {text}")
])
```

```python
#Task-6{to generate post}3defiing again to understand the flow in the chain invoke .. (no need to define multiple times)

llm_connect = ChatOpenAI(model="gpt-5-mini",
temperature=0,
api_key=os.environ["OPENAI_API_KEY"],
base_url="https://alertmanager.tecnicasreunidas.es/v1",
http_client=httpx.Client(verify=False)
)
```

```python
#Task-7 {output str parser} --> definiing again to make understand

str_parser = StrOutputParser()
```

```python
#Chain

chain = primary_prompt | llm_connect | str_parser | dict_maker_runnable | post_prompt | llm_connect | str_parser
chain.invoke("Overviw on  Telangana")
```
