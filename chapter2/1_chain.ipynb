```python
import os
import httpx
from dotenv import load_dotenv
from langchain_openai import OpenAI,ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm_connect = ChatOpenAI(model="gpt-5-mini",
                    temperature=0,
                    api_key=os.environ["OPENAI_API_KEY"],
                    base_url="https://alertmanager.tecnicasreunidas.es/v1",
                    http_client=httpx.Client(verify=False))

primary_prompt = ChatPromptTemplate.from_messages([
    ("system","You are professional assistant"),
    ("user","{input}")
])

user_input = input("Ask LLM Anything here..")
final_prompt = primary_prompt.invoke({"input":user_input})

llm_answer = llm_connect.invoke(final_prompt.messages).content
llm_answer
```

# Now lets make a Chain 

```python
# Task-1
from langchain_core.prompts import ChatPromptTemplate

primary_prompt=ChatPromptTemplate([
    ("system","You are a friendly assistant"),
    ("user","{input}")
])
```

```python
# Task-2
from langchain_openai import ChatOpenAI
import os
import httpx
from dotenv import load_dotenv

llm_connect1 = ChatOpenAI(model="gpt-5-mini",
                    temperature=0,
                    api_key=os.environ["OPENAI_API_KEY"],
                    base_url="https://alertmanager.tecnicasreunidas.es/v1",
                    http_client=httpx.Client(verify=False))
```

```python
# Task-3

from langchain_core.output_parsers import StrOutputParser #to parse the output of the llm
str_parser = StrOutputParser()
```

### ** Manual Invocation

```python
final_prompt = primary_prompt.invoke({"input":"What area the capitas of Andhrapradesh and Telanagana states"})
ans = llm_connect1.invoke(final_prompt)
ans.content
```

## Chain invocation

* using Pipeline operator

```python
chain = primary_prompt | llm_connect1 | str_parser #using a pipe line for series of tasks a simplest way to create a chain pipeline
chain.invoke({"input":"what is the capital of Telangana"})
```

* using RunnableSequence

```python
from langchain_core.runnables import RunnableSequence

chain = RunnableSequence(primary_prompt,llm_connect1,str_parser)
chain.invoke("what is the capital of Andhrapradesh")
```
